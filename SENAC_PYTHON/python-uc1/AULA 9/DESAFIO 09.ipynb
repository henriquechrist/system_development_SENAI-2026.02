## 1️⃣ Soma de Elementos


```python
produtos = (("Arroz",5.99,10),
            ("Feijão",7.49,4),
            ("Leite",4.89,5),
            ("Óleo",9.99,6),
            ("Açúcar",3.29,7))
valor_total=0
valor_produto=0
print("Lista de Produtos")

for nome,preco,qtd in produtos:
    valor_produto= preco*qtd
    valor_total+=valor_produto
    print(f"{nome:.<20} R${preco:.2f}..qtd{qtd:02d} ---- Valor: R${preco*qtd:.2f}")
print(f"\n\tValor Total da Compra: R$ {valor_total:7.2f}")
```

## 2️⃣ Primos

```python
numero=(2,7,10,13,17,22,29,31,40,53)

def eh_primo(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

primo_lista = []
for num in numero:
    if eh_primo(num):
        primo_lista.append(num)

primos = tuple(primo_lista)
print(f"Números primos:{primos}")
```

## 3- Classificação de idade

```python
idades=(18,25,16,30,14)

maiores=0
for idade in idades:
    if idade>=18:
        maiores += 1
print(f"Quantidade de Maiores de idade:{maiores}")
```
