## Estruturas Condicionais

```python
nome=input("Qual é seu nome?")
anonas=input(f"{nome}, qual ano você nasceu")
anoatu=2026

print(f"Seu nome é {nome}, você nasceu em {anonas}, o ano atual é {anoatu}")
```

```python
anosog=int(anonas)
idade=anoatu-anosog
if idade>=18:
    print(f"Você é maior de idade, tem {idade} anos")
elif idade>=16:
    print(f"Você é adolescente, tem {idade} anos")    
else:
    print(f"Você é menor de idade, tem {idade} anos")
```
