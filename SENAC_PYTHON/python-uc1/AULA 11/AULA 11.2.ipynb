## Criando dicionários com chaves e um valor padrão

```python
chaves = ["nome","dt_nascimento","cep","logradouro","bairro","cidade","estado"]
valor_padrão="N/A"
usuario = dict.fromkeys(chaves,valor_padrão)

print(f"{usuario}",end=" ,")
```

```python
print("Sistema - Sistemático de Análise Sistemas")

usuario["nome"]=input("Qual o seu nome?? ")
usuario["dt_nascimento"]=input("Digite sua data de nascimento(dd/mm/aaaa): ")
usuario["cep"]=input("Qual o seu cep?? ")
usuario["logradouro"]=input("Qual o seu logradouro?? ")
usuario["bairro"]=input("Qual o seu bairro?? ")
usuario["cidade"]=input("Qual a sua cidade?? ")
usuario["estado"]=input("Qual o seu estado?? ")


print(usuario)
```

```python
print(f"Seu nome é: {usuario['nome']}")
```

```python
for chave,valor in usuario.items():
    print(f"{chave.capitalize():.<20}: {valor.capitalize()}")
```

```python
clientes={}
clientes["12345678912"]=usuario
print(clientes)
```
