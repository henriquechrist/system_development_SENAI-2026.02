```python
# Definindo uma classe chamada 'Carro'
class Carro:
    # Método construtor (__init__) para inicializar os atributos
    def __init__(self, marca, modelo,  cor, ano="2026"):
        self.marca = marca      # Atributo
        self.modelo = modelo    # Atributo
        self.ano = ano          # Atributo
        self.ligado = False     # Atributo com valor padrão
        self.cor= cor
        self.tração=False
        self.transmissão=False
        self.gas = 0

    # Método para ligar o carro
    def ligar(self):
        if not self.ligado:
            self.ligado = True
            print(f"{self.marca} {self.modelo} está ligado.")
        else:
            print(f"{self.marca} {self.modelo} já está ligado.")

    # Método para desligar o carro
    def desligar(self):
        if self.ligado:
            self.ligado = False
            print(f"{self.marca} {self.modelo} está desligado.")
        else:
            print(f"{self.marca} {self.modelo} já está desligado.")

     # Método para tracionar o carro
    def Tracionar(self):
        if not self.tração:
            self.tração = True
            print(f"{self.marca} {self.modelo} tracionou.")
        else:
            print(f"{self.marca} {self.modelo} já está tracionado.")

    # Método para Destracionar o carro
    def Destracionar(self):
        if self.tração:
            self.tração = False
            print(f"{self.marca} {self.modelo} destracionou.")
        else:
            print(f"{self.marca} {self.modelo} já está destracionado.")

              # Método para tracionar o carro
    def Automatizar(self):
        if not self.transmissão:
            self.transmissão = True
            print(f"{self.marca} {self.modelo} tracionou.")
        else:
            print(f"{self.marca} {self.modelo} já está tracionado.")

    # Método para Destracionar o carro
    def Manual(self):
        if self.transmissão:
            self.transmissão = False
            print(f"{self.marca} {self.modelo} destracionou.")
        else:
            print(f"{self.marca} {self.modelo} já está destracionado.")
    
    def abastecer(self):
        if self.gas==0:
            meu_carro.gas=litragem
        print(f"Você abasteceu {litragem} litros")

    def zerar_combus(self):
        if self.gas>=0:
            meu_carro.gas=0

    # Método para exibir informações do carro
    def exibir_info(self):
        status = "ligado" if self.ligado else "desligado"
        print(f"{self.marca} {self.modelo} {self.ano}, de cor {self.cor} está {status}.")
```
