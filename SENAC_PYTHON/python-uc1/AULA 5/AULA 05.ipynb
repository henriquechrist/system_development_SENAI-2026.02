## CALCULADORA

```python
n1= float(input("Insira o primeiro número"))
print(f"O Primeiro Número é: {n1}")
op= input("Insira a operação: +; -; *; /")
n2= float(input("Insira o segundo número: "))
print(f"O Segundo número é: {n2}")
print (f"{n1} {op} {n2}")
if op == "+":
    print(n1+n2)
elif op == "-":
    print(n1-n2)
elif op == "*":
    print(n1*n2)
elif op =="/":
    print(n1/n2)
else: "Operação inválida"
```

## Calculadora IMC

```python
print("CALCULADORA IMC")
altura= float(input("Digite sua altura?: "))
peso= float(input("Digite seu peso"))
print(f"Seu peso é: {peso}")
print(f"Sua altura é: {altura}")
IMC= (peso/(altura**2))
print(f"Seu IMC é de: {IMC:.2f}")
      
```

## Desconto em Compras

```python
val=float(input("Digite o valor total da compra: "))
percentil=float(input("Qual a %? (insira somente número)"))
desc=(val*percentil)/100
print(f"O valor original é R$ {val}, o desconto é R$ {desc}, equivalentes a {percentil}%")
print(f"O valor final é R$ {val - desc}")
```

## Tempo de viagem

```python
distancia=float(input("Insira a distância em km: "))
velocidade=float(input("Insira a velocidade média em km/h: "))
print(f"A distância é {distancia} km, a velocidade média é {velocidade} km/h")
tempo= distancia/velocidade
print(f"Vai levar {tempo:.2f} minutos")
```

## IMC - com indicativo

```python
peso=float(input("digite o peso:"))
altura=float(input(" digite o altura: "))
imc=peso/(altura**2)
print(f"Sua altura é {altura} mts e seu peso {peso} kg")
if imc <18.5:
    print(f"Seu imc é {imc:.2f}, vocé está abaixo do peso ")
elif imc <=14.9:
    print(f"Seu imc é {imc:.2f}, vocé esta no peso ideal ")
elif imc <=29.9:
    print(f"Seu imc é {imc:.2f}, vocé esta sobre peso ")
elif imc >30:
    print(f"Seu imc é {imc:.2f}, vocé esta obeso ")
else:
    print("Operação Inválida")
```

## Ativida laço de repetição

```python
print(" A | B | A AND B ")
print("---|---|--------")

for A in [0,1]:
    for B in [0,1]:
        resultado = A and B 
        print(f"{A} | {B} |   {resultado}")
```
