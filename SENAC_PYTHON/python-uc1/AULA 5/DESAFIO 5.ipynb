## Desafio 2 - Aprovado na Faculdade

```python
prova1=float(input("Digite a nota da primeira prova: "))
prova2=float(input("Digite a nota da segunda prova: "))
trabalho=input("O aluno fez trabalho? (S/N)")
média=(prova1+prova2)/2
if média>=7 or trabalho=="S":
    print("Aprovado!")
else:
    print("Reprovado!")
```

## Desafio 3 - Controle de Entrada em uma Festa 

```python
nome= input("Qual o seu Nome?")
idade=int(input("Qual a sua idade? "))
banidos= ["João","Maria","Jonas"]
pais=input("Está acompanhado? (S/N)")
if idade >= 18 or pais == "S":
    print("Aguarde Verificação da Lista: ")
    if nome in banidos:
        print("Acesso Negado!")
    else:
        print("Entrada Liberada!")
```

## Desafio 4 - Jogo de Adivinhação FOR

```python
import random
segredo=random.randint(1,10)
for i in range(3):
    numero=int(input("Digite um numero:"))
    if numero == segredo:
        print("Parabéns, você acertou.")
        break
print(f"O número secreto era {segredo}")
```

## Desafio 4 - Jogo de adivinhação while

```python
import random
segredo=random.randint(0,10)
tentativas=0
while tentativas<3:
    numero=int(input("Digite um numero: "))
    if numero== segredo:
        print("Parabéns! Você acertou!")
        break
    else:
        print("Você errou, tente novamente")
    tentativas=tentativas+1
print(f"O número secreto era {segredo}.")
```
