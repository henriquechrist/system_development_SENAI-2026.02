## Exercicio 3

```python
def par_ou_impar(n):
    if n % 2 == 0:
        return "Par"
    else:
        return "Ímpar"
```

```python
print(f"O número 4 é {par_ou_impar(4)}")
print(f"O número 7 é {par_ou_impar(7)}")
```

## Exercicio 4

```python
def fatorial(b):
    if b <0:
        return "Numero Inválido"
    elif b == 0:
        return 1
    else:
        resultado = 1
        for i in range(1,b+1):
            resultado *= i
        return resultado
#Ex de chamada
print(f"Fatorial de 5: {fatorial(5)}")
```

## Exercicio 7 

```python
def eh_palindromo(palavra):
    return palavra == palavra[::-1]
print(f"ana é palíndromo? {eh_palindromo("ana")} ")
print(f"python é palíndromo? {eh_palindromo("python")} ")
```

## Exercício 8

```python
def contar_vogais(texto):
    vogais = "aeiouAEIOU"
    contador = 0
    for char in texto:
        if char in vogais:
            contador += 1 
    return contador

print(contar_vogais("Henrique"))
```

## Exercicio 9

```python
def tabuada(n):
    resultado = []
    for i in range(1,11):
        resultado.append(n*i)
    return resultado
print(tabuada(5))
```

## Exercicio 10

```python
def ordenar_por_segundo(lista_tuplas):
    return sorted(lista_tuplas, key=lambda x: x[1])
# Exemplo de chamada
lista_exemplo = [("a", 3), ("b", 1), ("c", 2)]
print("Lista ordenada:", ordenar_por_segundo(lista_exemplo))
```
