## Exercicio 5

```python
def fatorial(n):
   if n == 0 or n == 1:
       return 1
   else:
       return n * fatorial(n - 1)

def main():
   num = int(input("Digite um número: "))
   print(f"O fatorial de {num} é {fatorial(num)}")

if __name__ == "__main__":
   main()
```

```python
def fibonacci(n):
   sequencia = []
   a, b = 0, 1
   while a <= n:
       sequencia.append(a)
       a, b = b, a + b
   return sequencia

def main():
   num = int(input("Digite um número: "))
   print(f"Sequência de Fibonacci até {num}: {fibonacci(num)}")

if __name__ == "__main__":
   main()
```
