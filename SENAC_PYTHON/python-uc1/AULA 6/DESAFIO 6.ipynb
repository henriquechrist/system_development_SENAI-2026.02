```python
for i in range(10):
    i+=1
    soma=i*i
    print(f"O quadrado de {i:02d} é: {soma:03d}")
```

## 3- IMPAR PAR COM FOR

```python
for i in range(1,21):
    if i%2==0:
        print(f"O número {i:2d} é par")
    else:
        print(f"O número {i:2d} é impar")
```

```python
pares=[]
impares=[]
for i in range(1,21):
    if i%2==1:
       pares.append(i)
    else:
        impares.append(i)

print(f"NUMEROS PARES: {pares}")
print(f"NUMEROS IMPARES: {impares}")
```
