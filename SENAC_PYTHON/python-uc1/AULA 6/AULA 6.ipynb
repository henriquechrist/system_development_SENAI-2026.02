## Laços de repetição for 

```python
frutas = ["maça","banana","cereja"]
for fruta in frutas:
    print(fruta)
```

```python
for i in range(5):
    print(f"{i} - fruta - {i+1} ")
```

# Laços de repetição while

```python
contador = 0
while (contador <= 10):
    print(contador)
    contador+=1
```

## Exemplo matriz com for

```python
matriz = [
    [1,2,3],
    ["Eduardo","João","Maria"],
    [7.1,8.2,9.3]
]

for linha in matriz:
    for elemento in linha:
        print (f"{elemento}, ",end="\n")
    print(f"Olá Bom dia",end="\n""Você está bem?")
```

# Quebrar laços

```python
for i in range(10):
    if i == 10:
        break   #Para o loop i for 5
    if i % 2 == 0:
        continue # Pula números pares
    print(i)
```

# Else em loops

```python
for i in range(3):
    print(i)
else:
    print("Loop concluído")
```
