```python
import random

# Gera uma matriz 3x4 com números aleatórios de 1 a 99
linhas = 3
colunas = 4

matriz = []
for _ in range(linhas):
    linha = []
    for _ in range(colunas):
        linha.append(random.randint(1, 99))
    matriz.append(linha)

# Exibe a matriz original formatada
print("--- Matriz Original ---")
for linha in matriz:
    for elemento in linha:
        print(f"{elemento:3d}", end=' ')
    print()

# Cria a matriz transposta (invertendo linhas e colunas)
transposta = []
for c in range(colunas):
    nova_linha = []
    for l in range(linhas):
        nova_linha.append(matriz[l][c])
    transposta.append(nova_linha)

# Exibe a matriz transposta formatada
print("\n--- Matriz Transposta ---")
for linha in transposta:
    for elemento in linha:
        print(f"{elemento:3d}", end=' ')
    print()
    
```
